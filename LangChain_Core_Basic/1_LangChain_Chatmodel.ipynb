## Chat model 호출

 Chat LLM 사용 방법

대화형 LLM을 사용하기 위해 ChatOpenAI 클래스를 사용할 수 있습니다. 이 예제에서는 gpt-4.1-nano 모델을 사용합니다.

```python
# 1. 필수 라이브러리 가져오기 
import openai  # OpenAI 라이브러리 임포트
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

client = openai.OpenAI()  # OpenAI 클라이언트 생성
print("API 키 설정이 완료되었습니다.")  # API 키 설정 완료 확인

# Chat LLM(언어 모델) 객체 생성
# - model: 사용할 모델 (gpt-4.1-nano)
# - temperature: 응답의 다양성 조절 (값이 높을수록 다양한 답변 생성)
# - max_tokens: 생성할 응답의 최대 토큰 수 (응답의 길이를 조절)
chat_llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)

# 사용자 질문에 대한 응답 요청
question = "한국에서 가장 인기있는 장소의 그 장소에 대해 설명해 주세요."

# invoke 메서드로 질문에 대한 답변 생성
response = chat_llm.invoke(question)

# 생성된 답변 출력
print(response)
```

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage  # 메시지 타입 가져오기
# 1. 필수 라이브러리 가져오기 
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기
model = ChatOpenAI(model="gpt-4.1-nano")  # LangChain ChatOpenAI 모델 생성
conversation = [
    SystemMessage("당신은 한국어를 영어로 번역하는 도움이 되는 어시스턴트입니다."),  # 시스템 역할 설정
    HumanMessage("번역해주세요: 나는 프로그래밍을 좋아합니다."),  # 사용자 첫 번째 질문
    AIMessage("I love programming."),  # AI 답변 예시 (few-shot)
    HumanMessage("번역해주세요: 나는 애플리케이션 만드는 것을 좋아합니다.")  # 사용자 두 번째 질문
]

# 모델에 대화 내용을 전달하여 응답 생성
response = model.invoke(conversation)

# 응답 내용만 출력 (.content = 텍스트만 추출)
print(response.content)  # I love building applications.
```

## Stream ChatOpenAI

https://docs.langchain.com/oss/python/langchain/models#stream


```python
# 1. 필수 라이브러리 가져오기 
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

model = ChatOpenAI(model="gpt-4.1-nano")  # LangChain ChatOpenAI 모델 생성

# 스트리밍 방식으로 실시간 출력
# end="" : 줄바꿈 없이 이어서 출력
# flush=True : 버퍼 즉시 비워서 실시간으로 화면에 표시
for chunk in model.stream("하늘은 무슨 색인가요?"):
    print(chunk.text, end="", flush=True)

# 출력 예시:
# 하늘은 일반적으로 파란색입니다. 이는 레일리 산란...
```

```python
# 1. 필수 라이브러리 가져오기 
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

model = ChatOpenAI(model="gpt-4.1-nano")  # LangChain ChatOpenAI 모델 생성

# 스트리밍 방식으로 모델 응답 받기
# model.stream() : 응답을 토큰 단위로 조금씩 받아옴
full = None  # None | AIMessageChunk - 누적된 전체 메시지 저장 변수

for chunk in model.stream("하늘은 무슨 색인가요?"):
    # 첫 번째 청크면 그대로 저장, 이후엔 이전 청크에 누적(+)
    full = chunk if full is None else full + chunk
    
    # 현재까지 누적된 텍스트 출력 (토큰이 쌓일수록 문장이 완성됨)
    print(full.text)

# 출력 예시:
# 하늘
# 하늘은
# 하늘은 일반적으로
# 하늘은 일반적으로 파란색
# 하늘은 일반적으로 파란색입니다
# ...

print("=" * 30)

# content_blocks : 최종 완성된 메시지를 블록 구조로 확인
# type: 콘텐츠 타입 (text, image 등)
# text: 실제 텍스트 내용
print(full.content_blocks)
# [{"type": "text", "text": "하늘은 일반적으로 파란색입니다..."}]
```
