## Basic usage

```python
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, AIMessage, SystemMessage

# gpt-5-nano 모델 초기화
model = init_chat_model("gpt-5-nano")

# 시스템 메시지: AI 역할 및 행동 지침 설정
system_msg = SystemMessage("당신은 친절한 한국어 어시스턴트입니다.")

# 휴먼 메시지: 사용자 입력
human_msg = HumanMessage("안녕하세요, 어떻게 지내세요?")

# 메시지 리스트 구성 후 모델 호출 → AIMessage 반환
messages = [system_msg, human_msg]
response = model.invoke(messages)

# 모델 응답 출력
print(response.content)
```

## Message types

### System message 
Basic instructions


```python
from langchain.messages import SystemMessage, HumanMessage
from langchain.chat_models import init_chat_model
# gpt-5-nano 모델 초기화
model = init_chat_model("gpt-5-nano")

# 시스템 메시지: AI 역할 설정 (코딩 어시스턴트)
system_msg = SystemMessage("당신은 친절한 파이썬 코딩 전문가입니다.")

# 대화 메시지 리스트 구성
messages = [
    system_msg,
    # 사용자 질문: REST API 생성 방법
    HumanMessage("파이썬으로 REST API를 만드는 방법을 간단하게 알려줘")
]

# 모델 호출 → AIMessage 반환
response = model.invoke(messages)

# 모델 응답 출력
print(response.content)
```

### AI message

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage
from langchain.chat_models import init_chat_model
# gpt-5-nano 모델 초기화
model = init_chat_model("gpt-5-nano")

# AI 메시지 수동 생성 (대화 히스토리 구성용)
ai_msg = AIMessage("물론이죠, 질문에 대해 기꺼이 도와드리겠습니다!")

# 대화 히스토리 구성
messages = [
    # 시스템 메시지: AI 역할 설정
    SystemMessage("당신은 친절한 어시스턴트입니다."),
    
    # 1번째 턴: 사용자 질문
    HumanMessage("도와주실 수 있나요?"),
    
    # 1번째 턴: AI 이전 응답 (수동 삽입)
    ai_msg,
    
    # 2번째 턴: 사용자 후속 질문
    HumanMessage("좋아요! 2+2는 얼마인가요?")
]

# 전체 대화 히스토리를 포함하여 모델 호출
response = model.invoke(messages)

# 모델 응답 출력
print(response.content)
```

Tool calls

```python
from langchain.chat_models import init_chat_model

# gpt-4o-mini 모델 초기화
model = init_chat_model("gpt-4o-mini")

# 날씨 조회 도구 함수 정의
def get_weather(location: str) -> str:
    """특정 위치의 날씨를 조회합니다."""
    # 실제 구현 시 날씨 API 호출 (예: OpenWeatherMap)
    return f"{location}의 현재 날씨는 맑음, 기온 23°C입니다."

# 모델에 도구 바인딩
model_with_tools = model.bind_tools([get_weather])

# 모델 호출 → 도구 호출 여부 판단 후 AIMessage 반환
response = model_with_tools.invoke("파리의 날씨는 어때?")

# AIMessage에 포함된 tool_calls 순회
for tool_call in response.tool_calls:
    print(f"도구 이름: {tool_call['name']}")   # 호출할 도구명
    print(f"인자값:   {tool_call['args']}")    # 도구에 전달할 인자
    print(f"호출 ID:  {tool_call['id']}")      # ToolMessage 연결용 고유 ID
```

Streaming and chunks

```python
from langchain.chat_models import init_chat_model

# gpt-4o-mini 모델 초기화
model = init_chat_model("gpt-4o-mini")
# 스트리밍 응답 청크 저장 리스트
chunks = []
full_message = None

# 모델 스트리밍 호출 → 응답을 청크 단위로 실시간 수신
for chunk in model.stream("안녕하세요!"):
    chunks.append(chunk)                  # 청크 리스트에 저장
    print(chunk.text, end="", flush=True) # 실시간 출력 (줄바꿈 없이)
    
    # 청크를 누적하여 전체 메시지 조합
    full_message = chunk if full_message is None else full_message + chunk

print()  # 스트리밍 완료 후 줄바꿈

# 누적된 전체 메시지 출력
print(f"\n전체 메시지: {full_message.content}")
print(f"토큰 사용량: {full_message.usage_metadata}")
```

### Tool message

```python
from langchain.messages import AIMessage, ToolMessage, HumanMessage
from langchain.chat_models import init_chat_model

# gpt-4o-mini 모델 초기화
model = init_chat_model("gpt-4o-mini")

# ✅ 1단계: AI가 도구 호출을 결정한 메시지 수동 생성
ai_message = AIMessage(
    content=[],                    # 도구 호출 시 텍스트 응답은 비어있음
    tool_calls=[{
        "name": "get_weather",     # 호출할 도구명
        "args": {"location": "서울"},  # 도구에 전달할 인자
        "id": "call_123"           # 고유 호출 ID (ToolMessage와 반드시 매핑)
    }]
)

# ✅ 2단계: 도구 실행 및 결과 생성
weather_result = "맑음, 기온 23°C"

tool_message = ToolMessage(
    content=weather_result,        # 도구 실행 결과
    tool_call_id="call_123"        # AIMessage의 호출 ID와 반드시 일치해야 함
)

# ✅ 3단계: 전체 대화 히스토리 구성
messages = [
    HumanMessage("서울 날씨가 어때?"),  # 사용자 질문
    ai_message,                          # AI의 도구 호출 결정
    tool_message,                        # 도구 실행 결과
]

# ✅ 4단계: 도구 결과를 반영한 최종 응답 생성
response = model.invoke(messages)
print(response.content)
```
