# 프롬프트 템플릿(Prompt Template)

Langchain의 Prompt Template 기능은 사용자 입력을 변수화하여 동적이고 반복 가능한 프롬프트를 생성하는 데 사용됩니다.

LangChain 라이브러리의 `PromptTemplate`과 `ChatPromptTemplate`을 사용할 수 있습니다. 이 코드에서는 이러한 템플릿을 이용해 간단한 프롬프트를 생성하는 방법을 설명합니다.


### 기본적인 프롬프트 템플릿 생성

 사용자가 제공한 입력을 바탕으로 간단한 프롬프트를 생성

```python
# 필요한 라이브러리 임포트
from langchain_core.prompts import PromptTemplate  # 프롬프트 템플릿 생성을 위한 라이브러리
from langchain_core.prompts import ChatPromptTemplate  # 챗봇용 프롬프트 템플릿 생성을 위한 라이브러리
```

```python
# 용어를 쉽게 설명하는 챗봇 템플릿
simple_explain_template = "당신은 초등학생에게 {term}을(를) 쉽게 설명하는 챗봇입니다."
print(simple_explain_template)  # 템플릿을 출력하여 확인

# 기본 파이썬 라이브러리에서 .format()을 사용하여 템플릿을 활용하는 방법
# 'term'이라는 자리 표시자에 '인공지능'을 삽입하여 템플릿을 포맷
formatted_template = simple_explain_template.format(term='인공지능')
print(formatted_template)  # 포맷된 템플릿 출력
```

### openai 연결

```python
# 1. 필수 라이브러리 가져오기 
import openai  # OpenAI 라이브러리 임포트
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

client = openai.OpenAI()  # OpenAI 클라이언트 생성
print("API 키 설정이 완료되었습니다.")  # API 키 설정 완료 확인

# Chat LLM(언어 모델) 객체 생성
# - model: 사용할 모델 (gpt-4.1-nano)
# - temperature: 응답의 다양성 조절 (값이 높을수록 다양한 답변 생성)
# - max_tokens: 생성할 응답의 최대 토큰 수 (응답의 길이를 조절)
llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)

# LangChain의 PromptTemplate을 사용하여 프롬프트 생성
# input_variables: 템플릿에서 사용할 변수 (여기서는 'term')
# template: 미리 정의한 챗봇 템플릿
explain_prompt = PromptTemplate(input_variables=["term"],
                                template=simple_explain_template)

# 프롬프트 생성 및 실행 예시
# 'term' 변수에 '인공지능'을 넣어 템플릿을 포맷
formatted_prompt = explain_prompt.format(term="인공지능")
response = llm.invoke(formatted_prompt)  # LLM 모델을 사용하여 응답 생성
print(response)  # 생성된 응답 출력
```

### openai & PromptTemplate 

```python
from langchain_core.prompts import PromptTemplate 
# 1. 필수 라이브러리 가져오기 
import openai  # OpenAI 라이브러리 임포트
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

client = openai.OpenAI()  # OpenAI 클라이언트 생성
print("API 키 설정이 완료되었습니다.")  # API 키 설정 완료 확인
# Chat LLM(언어 모델) 객체 생성
# - model: 사용할 모델 (gpt-4.1-nano)
# - temperature: 응답의 다양성 조절 (값이 높을수록 다양한 답변 생성)
# - max_tokens: 생성할 응답의 최대 토큰 수 (응답의 길이를 조절)
llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)

# 특정 템플릿 문자열을 사용하여 PromptTemplate 인스턴스를 생성합니다
# 'product'라는 자리 표시자를 포함하는 템플릿 문자열
prompt = PromptTemplate(
    template="{product}는 어느 회사에서 개발한 제품인가요？",
    input_variables=["product"]  # 템플릿에서 사용될 변수 'product'를 정의
)
# 'prompt' 객체를 사용하여 템플릿을 다양한 제품 이름으로 포맷하고 결과를 출력합니다
print(llm.invoke(prompt.format(product="아반떼")))  # '아반떼'로 포맷
print(llm.invoke(prompt.format(product="소나타")))  # '소나타'로 포맷
print(llm.invoke(prompt.format(product="벤츠")))  # '벤츠'로 포맷
```

### 두 개의 매개변수를 사용하는 프롬프트 생성 1

여러 매개변수를 받는 프롬프트: 특정 주제에 대해 원하는 언어로 설명하는 프롬프트를 생성


```python
from langchain_core.prompts import PromptTemplate 
# 1. 필수 라이브러리 가져오기 
import openai  # OpenAI 라이브러리 임포트
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

client = openai.OpenAI()  # OpenAI 클라이언트 생성
print("API 키 설정이 완료되었습니다.")  # API 키 설정 완료 확인

# Chat LLM(언어 모델) 객체 생성
# - model: 사용할 모델 (gpt-4.1-nano)
# - temperature: 응답의 다양성 조절 (값이 높을수록 다양한 답변 생성)
# - max_tokens: 생성할 응답의 최대 토큰 수 (응답의 길이를 조절)
llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)

# 번역을 위한 프롬프트 템플릿
translate_template = "{topic}에 대해 {language}로 설명해 주세요."

# PromptTemplate으로 템플릿 생성
translate_prompt = PromptTemplate(input_variables=["topic", "language"],
                                  template=translate_template)

# 프롬프트 생성 및 실행 예시
formatted_translate_prompt = translate_prompt.format(topic='컴퓨터 비전', language='한국어')
response_translate = llm.invoke(formatted_translate_prompt)

print(response_translate)   # 생성된 번역 응답 출력
```

# 채팅 메시지를 위한 ChatPromptTemplate 생성

채팅 대화의 흐름을 설정하여 프롬프트를 생성합니다. 시스템 메시지와 사용자 메시지를 순차적으로 연결해 자연스러운 대화를 유도



```python
from langchain_core.prompts import  ChatPromptTemplate
# 1. 필수 라이브러리 가져오기 
import openai  # OpenAI 라이브러리 임포트
from langchain_openai import ChatOpenAI  # LangChain 라이브러리에서 ChatOpenAI 가져오기

client = openai.OpenAI()  # OpenAI 클라이언트 생성
print("API 키 설정이 완료되었습니다.")  # API 키 설정 완료 확인

# Chat LLM(언어 모델) 객체 생성
# - model: 사용할 모델 (gpt-4.1-nano)
# - temperature: 응답의 다양성 조절 (값이 높을수록 다양한 답변 생성)
# - max_tokens: 생성할 응답의 최대 토큰 수 (응답의 길이를 조절)
llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)
# ChatPromptTemplate을 이용해 채팅 대화 흐름 정의
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", '당신은 모든 질문에 대해 긍정적인 답변을 하는 챗봇입니다.'),  # 시스템 메시지: 챗봇의 역할과 성격을 정의
    ("user", 'LangChain을 배우면 어떤 유용한 점이 있나요?')  # 사용자 메시지: 질문 내용
])

# 메시지를 포맷팅하여 LLM에 전달
formatted_chat_messages = chat_prompt.format_messages()  # 대화 내용을 LLM이 이해할 수 있는 형식으로 정리
print(formatted_chat_messages)  

response_chat = llm.invoke(formatted_chat_messages)  # 정리된 메시지를 LLM에 전달하여 응답 생성
print(response_chat.content)  # 응답 출력
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI  # ChatOpenAI 임포트 (누락된 import 추가)
import openai  # OpenAI 라이브러리 임포트

client = openai.OpenAI()  # OpenAI 클라이언트 생성

# ChatOpenAI 모델 초기화
# - model: 사용할 GPT 모델 지정
# - temperature: 응답의 창의성 조절 (0=일관적, 1=창의적)
# - max_tokens: 응답 최대 토큰 수
llm = ChatOpenAI(model='gpt-4.1-nano', temperature=0.5, max_tokens=1024)

# ChatPromptTemplate 정의
# 대화 흐름을 역할(role)별로 구성:
#   - system : 모델의 페르소나/행동 방식 지정
#   - human  : 사용자 발화
#   - ai     : 이전 AI 응답 (few-shot 예시로 활용)
# {name}, {user_input} 은 나중에 채워질 변수 (placeholder)
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. Your name is {name}."),
        ("human", "Hello, how are you doing?"),
        ("ai", "I'm doing well, thanks!"),
        ("human", "{user_input}"),
    ]
)

# 템플릿에 실제 값을 주입하여 메시지 리스트 생성
# - name      → "Bob"  (시스템 프롬프트의 {name} 대체)
# - user_input → "What is your name?"  (마지막 human 메시지 대체)
prompt_value = template.invoke(
    {
        "name": "Bob",
        "user_input": "What is your name?",
    }
)

# 완성된 프롬프트 출력 (ChatPromptValue 객체 — 메시지 리스트 포함)
print(prompt_value)

# LLM에 프롬프트를 전달하여 응답 생성
response_chat = llm.invoke(prompt_value)

# AIMessage 객체에서 텍스트 내용만 추출하여 출력
print(response_chat.content)

```
